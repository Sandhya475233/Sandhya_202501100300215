```python
# Step 1: Import libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, roc_curve
import matplotlib.pyplot as plt

# Step 2: Load dataset (example CSV)
# Assume columns: ['diagnosis_code', 'vitals', 'prior_visits', 'readmitted']
data = pd.read_csv("hospital_data.csv")

# Step 3: Features and target
X = data[['diagnosis_code', 'vitals', 'prior_visits']]
y = data['readmitted']

# Step 4: Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Step 5: Logistic Regression with L2 regularization
model = LogisticRegression(penalty='l2', solver='liblinear')
model.fit(X_train, y_train)

# Step 6: Predictions (probabilities)
y_pred_prob = model.predict_proba(X_test)[:,1]

# Step 7: Evaluate with ROC-AUC
auc = roc_auc_score(y_test, y_pred_prob)
print("ROC-AUC Score:", auc)

# Step 8: Plot ROC curve
fpr, tpr, thresholds = roc_curve(y_test, y_pred_prob)
plt.plot(fpr, tpr, label="ROC curve")
plt.plot([0,1],[0,1],'--',color='gray')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()
```
